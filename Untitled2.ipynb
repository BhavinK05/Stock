```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Dense,Flatten

model = Sequential([
    Conv2D(32,(3, 3),activation='relu',input_shape=(28,28,1)),
    MaxPooling2D((2,2)),
    Flatten(),
    Dense(64,activation='relu'),
    Dense(10,activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print('CNN model created successfully')
model.summary()
```

```python
import os
import shutil

# 1. Export the Keras model to a temporary SavedModel directory
temp_model_path = "temp_keras_model"
if os.path.exists(temp_model_path):
    shutil.rmtree(temp_model_path)

# Use model.export for SavedModel format in newer Keras/TF versions
model.export(temp_model_path)

# 2. Use the tf2onnx command line tool to perform the conversion
!python -m tf2onnx.convert --saved-model temp_keras_model --output keras_model.onnx --opset 13

# 3. Clean up and verify
if os.path.exists("keras_model.onnx"):
    print("\nSuccess: Model converted and saved as 'keras_model.onnx'")
    shutil.rmtree(temp_model_path)
else:
    print("\nConversion failed.")
```
